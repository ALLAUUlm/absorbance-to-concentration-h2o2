```python
from collections import defaultdict
import re

output_dir = 'data/pre_metadata_processed'
os.makedirs(output_dir, exist_ok=True)

csv_file_mapping = []
batch_counts = defaultdict(int)
last_spot_for_material = defaultdict(lambda: None)
existing_batches_for_material_spot = defaultdict(set)  # NEW: track all batches for (material, spot)
measurement_counts = defaultdict(lambda: defaultdict(int))  # batcho -> spot -> count

for csv_file in oldraw_files:
    df = pd.read_csv(csv_file)
    base = os.path.basename(csv_file)
    parts = base.split('_')
    date = parts[0]
    material = next((p for p in parts if p.startswith('LN') or p.startswith('BLPHI')), "UNKNOWN")

    # First, skip repeated measurements and clean up the column list
    filtered_indices = []
    i = 0
    while i < len(df.columns) - 1:
        col1 = df.columns[i]
        if i + 2 < len(df.columns) - 1:
            next_col1 = df.columns[i + 2]
            base_col1 = re.sub(r'_r.*$', '', next_col1)
            curr_base_col1 = re.sub(r'_r.*$', '', col1)
            if base_col1 == curr_base_col1 and next_col1 != col1:
                print(f'the measurement {curr_base_col1} was measured again')
                i += 2  # skip the repeated pair
                continue
        filtered_indices.append(i)
        i += 2

    # Now create the spot list only for the filtered measurements
    spots = []
    loading = None
    for idx in filtered_indices:
        col1 = df.columns[idx]
        tokens = col1.split("_")
        spot = None
        for t_idx, token in enumerate(tokens):
            if token.replace("-", ".").replace(".", "").isdigit():
                loading = token.replace("-", ".")
                if t_idx + 1 < len(tokens):
                    spot_candidate = tokens[t_idx + 1]
                    if spot_candidate.isdigit():
                        spot = int(spot_candidate)
                break
        spots.append(spot)
    print(spots)
    if all(s is None for s in spots) and len(filtered_indices) == 4:
        spots = [1, 2, 3, 4]

    idx_spot = 0
    for idx, i in enumerate(filtered_indices):
        col1 = df.columns[i]
        col2 = df.columns[i + 1]
        spot = spots[idx_spot] if idx_spot < len(spots) else None

        # Extract irradiation time(s) for this measurement pair and print them with the csv file
        irradiation_times = []
        for col in [col1, col2]:
            match = re.search(r'(\d+(?:-\d+)?)(?=h)', col)
            if match:
                irradiation_times.append(match.group(1))
        irradiation_time = irradiation_times[0] if irradiation_times else None
        print(f"File: {csv_file}, Columns: {col1}, {col2}, Irradiation times: {irradiation_times}")

        # --- MODIFIED LOGIC FOR PER-CSV BATCHING ---
        batch_key = (material, spot, date)
        batch_exists = False
        for prev in csv_file_mapping:
            prev_material = prev['original_csv'].split('_')
            prev_material = next((p for p in prev_material if p.startswith('LN') or p.startswith('BLPHI')), "UNKNOWN")
            prev_spot = prev.get('spot', None)
            prev_new_csv = os.path.basename(prev['new_csv'])
            prev_tokens = prev_new_csv.split('_')
            prev_date = prev_tokens[0] if prev_tokens else None
            if (prev_material, prev_spot, prev_date) == batch_key:
                batch_exists = True
                prev_batcho_for_spot = prev_tokens[3] if len(prev_tokens) > 3 else None
                break

        if batch_exists:
            batcho = prev_batcho_for_spot
        else:
            batch_counts[material] += 1
            batcho = f"{material}o{batch_counts[material]:04d}"
            existing_batches_for_material_spot[(material, spot)].add(batcho)

        # Only extract data from col1 and col2 until value in col1 is in 199.5 <= float <= 200.5
        data_pair = df[[col1, col2]]
        mask = []
        data_end_found = False
        for v in data_pair[col1]:
            try:
                f = float(v)
                if 199.5 <= f <= 200.5:
                    mask.append(True)
                    data_end_found = True
                    break
            except Exception:
                pass
            mask.append(True)
        # After finding the value, include all rows up to and including that line, then stop
        if data_end_found:
            # Pad mask to length of data_pair
            mask += [False] * (len(data_pair) - len(mask))
        else:
            # If not found, keep all rows
            mask = [True] * len(data_pair)
        # Pad mask to length of data_pair
        mask += [False] * (len(data_pair) - len(mask))
        data_pair = data_pair[mask]

        irradiation_time = None
        for col in [col1, col2]:
            if "h" in col:
                irradiation_time = ''.join([c for c in col if c.isdigit()])
                break

        irradiation_map = {'0': 1, '1': 2, '2': 3, '3': 4, '4': 5, '6': 6, '24': 7}
        meas_num = irradiation_map.get(irradiation_time, measurement_counts[batcho][spot] + 1)
        measurement_counts[batcho][spot] += 1
# Here I want to also extract the metadata from the original CSV file

    # Find the metadata line that matches the column header of col1 and ends with the wavelength and a float
    # Find the index in meta_lines_clean where the wavelength drops below 200.5 (i.e., after the last data row)
    # Find the metadata block in meta_lines that starts with the first column header of new_csv and ends with the first line where the first value is a float < 200.5
    metadata_block = ""
    start_idx = None
    end_idx = None

    # Find the first column header of the new CSV (col1)
    first_col_header = col1.strip()
    # Ensure meta_lines is defined, otherwise fallback to an empty list
    try:
        meta_lines_local = meta_lines
    except NameError:
        meta_lines_local = []

    for idx, line in enumerate(meta_lines_local):
        if line.strip().startswith(first_col_header):
            start_idx = idx
            break

    # Find the end index: first line after start_idx where the first value is a float < 200.5
    if start_idx is not None:
        for idx in range(start_idx + 1, len(meta_lines_local)):
            line = meta_lines_local[idx]
            first_val = line.split(',')[0]
            try:
                if float(first_val) < 200.5:
                    end_idx = idx
                    break
            except Exception:
                continue

    # Extract the metadata block
    if start_idx is not None and end_idx is not None:
        metadata_block = "".join(meta_lines_local[start_idx:end_idx])
    elif start_idx is not None:
        metadata_block = "".join(meta_lines_local[start_idx:])

    # If found, append this metadata block to the new CSV file
    if metadata_block:
        with open(out_csv, 'a', encoding='utf-8') as f:
            f.write('\n' + metadata_block)

        out_csv = os.path.join(
            output_dir,
            f"{date}_UVV_ALLA_{batcho}_{meas_num}.csv"
        )
        data_pair.to_csv(out_csv, index=False)

        csv_file_mapping.append({
            'new_csv': out_csv,
            'original_csv': csv_file,
            'spot': spot,
            'first_column_header': col1,
            'original_csv_parts': base.split('_'),
            'new_csv_parts': os.path.basename(out_csv).split('_'),
            'first_column_header_parts': col1.split('_')
        })
        idx_spot += 1
# After processing all files, print all existing batches for each (material, spot):
print("All existing batches for each (material, spot):")
for key, batches in existing_batches_for_material_spot.items():
    print(f"{key}: {sorted(batches)}")
print(csv_file_mapping)
```

```python
# Extract snippets for each entry in csv_file_mapping
snippets_summary = []

for mapping in csv_file_mapping:
    # Extract file parts
    original_csv = mapping['original_csv']
    new_csv = mapping['new_csv']
    spot = mapping.get('spot', None)
    first_col = mapping.get('first_column_header', None)
    
    # Parse original csv filename
    orig_base = os.path.splitext(os.path.basename(original_csv))[0]
    orig_parts = orig_base.split('_')
    
    # Parse new csv filename
    new_base = os.path.splitext(os.path.basename(new_csv))[0]
    new_parts = new_base.split('_')
    
    # Extract date (usually first part)
    date = orig_parts[0] if len(orig_parts) > 0 else None
    
    # Extract material (LNxx or BLPHI)
    material = next((p for p in orig_parts if p.startswith('LN') or p.startswith('BLPHI')), None)
    
    # Extract batch from new csv (e.g., LN46o0001)
    batch = None
    for p in new_parts:
        if (p.startswith('LN') or p.startswith('BLPHI')) and 'o' in p:
            batch = p
            break
    
    # Extract irradiation time (look for something ending with 'h')
    irradiation_time = next((p.replace('h', '') for p in orig_parts if isinstance(p, str) and p.endswith('h')), None)
    
    # Extract wavelength (look for something ending with 'nm')
    wavelength = next((p.replace('nm', '') for p in orig_parts if isinstance(p, str) and p.endswith('nm')), None)
    
    # Extract loading (look for float-like string, e.g., '0-75' or '0.75')
    loading = next((p.replace('-', '.') for p in orig_parts if any(c.isdigit() for c in p) and ('.' in p or '-' in p)), None)
    
    # Extract synthesis date (last part, if 6 or 4 digits)
    synthesis_date = None
    for p in reversed(orig_parts):
        if len(p) == 6 and p.isdigit():
            # Convert DDMMYY to YYYYMMDD
            synthesis_date = f"20{p[4:6]}{p[2:4]}{p[0:2]}"
            break
        elif len(p) == 4 and p.isdigit():
            synthesis_date = p
            break
    
    # Use spot from mapping if available
    spot_val = spot
    
    snippets_summary.append({
        'date': date,
        'irradiation.time.value': irradiation_time,
        'system.wavelength.value': wavelength,
        'system.Loading.value': loading,
        'system.material.name': material,
        'system.material.synthesis date': synthesis_date,
        'spot': spot_val,
        'system.batch': batch,
    })

# Display first 5 for inspection
snippets_summary
```

```python
import re

# Load the demo template structure using ruamel.yaml to preserve formatting and indentation
yaml_ruamel = YAML()
with open(file[0], 'r') as f:
    demo_structure = yaml_ruamel.load(f)
    pprint.pprint(demo_structure)

# Explanation:
# In YAML, a list under a key (like extraction:) is represented with a dash (-) at the next indentation level.
# For example:
# extraction:
#   - repitition: 0
#     sample extracted:
#       value: 0.5
# Here, extraction: is a key whose value is a list. Each item in the list starts with a dash and is indented.
# When ruamel.yaml loads this, it creates a Python list under the 'extraction' key, where each list item is a dict.
# So, demo_structure['extraction'] is a list, and demo_structure['extraction'][0] is the first dict (with 'repitition', etc).

# When you fill in the extracted data, always assign to the first dict in the extraction list:
# structure['extraction'][0]['sample extracted']['value'] = ...
# structure['extraction'][0]['complexing agent']['added']['value'] = ...
# structure['extraction'][0]['sample dilution']['added'][0]['value'] = ...
# This ensures the output YAML will keep the extraction: - ... list-of-dicts structure.

# Note: ruamel.yaml should be used for both reading and writing to preserve formatting.
# Replace yaml.safe_dump with yaml_ruamel.dump for writing.
# with open(yaml_path, 'w', encoding='utf-8') as out_f:
#     yaml_ruamel.dump(structure, out_f)
# Ensure output directory exists
out_folder = 'data/pre_metadata_processed'
os.makedirs(out_folder, exist_ok=True)

for idx, (snip_dict, mapping) in enumerate(zip(snippets_summary, csv_file_mapping)):
    csv_filename = os.path.basename(mapping['new_csv'])
    # if csv_filename != '20240228_UVV_ALLA_BLPHIo0008_7.csv':
    #     continue  # Only process the specified file

    yaml_filename = os.path.splitext(csv_filename)[0] + '.yaml'
    yaml_path = os.path.join(out_folder, yaml_filename)

    structure = copy.deepcopy(demo_structure)

    # Fill standard fields
    if 'date' in snip_dict and snip_dict['date']:
        structure['date'] = snip_dict['date']
    if 'irradiation.time.value' in snip_dict and snip_dict['irradiation.time.value']:
        if 'irradiation' in structure and 'time' in structure['irradiation']:
            try:
                structure['irradiation']['time']['value'] = int(float(snip_dict['irradiation.time.value']))
            except Exception:
                structure['irradiation']['time']['value'] = snip_dict['irradiation.time.value']
    if 'system.wavelength.value' in snip_dict and snip_dict['system.wavelength.value']:
        if 'system' in structure and 'wavelength' in structure['system']:
            try:
                structure['system']['wavelength']['value'] = int(float(snip_dict['system.wavelength.value']))
            except Exception:
                structure['system']['wavelength']['value'] = snip_dict['system.wavelength.value']
    if 'system.Loading.value' in snip_dict and snip_dict['system.Loading.value']:
        if 'system' in structure and 'Loading' in structure['system']:
            try:
                structure['system']['Loading']['value'] = float(snip_dict['system.Loading.value'])
            except Exception:
                structure['system']['Loading']['value'] = snip_dict['system.Loading.value']
    if 'system.material.name' in snip_dict and snip_dict['system.material.name']:
        if 'system' in structure and 'material' in structure['system']:
            structure['system']['material']['name'] = snip_dict['system.material.name']
    if 'system.material.synthesis date' in snip_dict and snip_dict['system.material.synthesis date']:
        if 'system' in structure and 'material' in structure['system']:
            structure['system']['material']['synthesis date'] = snip_dict['system.material.synthesis date']
    if 'spot' in snip_dict and snip_dict['spot'] is not None:
        structure['spot'] = snip_dict['spot']

    # Adjust the batchname to the batch from the new_csv_file
    new_csv_base = os.path.basename(mapping['new_csv'])
    new_csv_parts = new_csv_base.split('_')
    batch_from_new_csv = None
    for part in new_csv_parts:
        if (part.startswith('LN') or part.startswith('BLPHI')) and 'o' in part:
            batch_from_new_csv = part
            break
    if batch_from_new_csv and 'system' in structure and 'batch' in structure['system']:
        structure['system']['batch'] = batch_from_new_csv

    # Adjust purging time based on irradiation time
    irr_time = snip_dict.get('irradiation.time.value')
    if irr_time is not None:
        try:
            irr_time_val = int(float(irr_time))
            if 'system' in structure and 'purging' in structure['system'] and 'time' in structure['system']['purging']:
                if irr_time_val in [0, 6]:
                    structure['system']['purging']['time']['value'] = 10
                    structure['system']['purging']['time']['unit'] = 'min'
                elif irr_time_val == 24:
                    structure['system']['purging']['time']['value'] = 0
                    structure['system']['purging']['time']['unit'] = 'min'
        except Exception:
            pass

    # --- Custom extraction from first_column_header_parts ---
    found = False
    header_parts = mapping.get('first_column_header_parts', [])
    for i, part in enumerate(header_parts):
        # sample extracted
        if i + 1 < len(header_parts) and header_parts[i + 1] == "s":
            val = part.replace("mL", "").replace("ml", "")
            try:
                val = float(val)
                # Add the extracted sample volume to the dict under 'sample extracted' -> 'value'
                if (
                    'extraction' in structure
                    and isinstance(structure['extraction'], list)
                    and len(structure['extraction']) > 0
                    and 'sample extracted' in structure['extraction'][0]
                    and isinstance(structure['extraction'][0]['sample extracted'], dict)
                ):
                    structure['extraction'][0]['sample extracted']['value'] = val
            except Exception:
                print(f"Error parsing sample extracted value: {val}")
                pass
        # complexing agent added
        if i + 1 < len(header_parts) and header_parts[i + 1] == "Ti":
            val = part.replace("mL", "").replace("ml", "")
            try:
                val = float(val)
                structure['extraction'][0]['complexing agent']['added']['value'] = val
            except Exception:
                pass
        # sample dilution added (for H2O)
        # sample dilution added (for H2O)
        if part == "H2O" and i > 0:
            val_str = header_parts[i - 1]
            if "mL" in val_str or "ml" in val_str:
                val = val_str.replace("mL", "").replace("ml", "")
                try:
                    val = float(val)
                    if (
                        'extraction' in structure
                        and isinstance(structure['extraction'], list)
                        and len(structure['extraction']) > 0
                        and 'sample dilution' in structure['extraction'][0]
                        and 'added' in structure['extraction'][0]['sample dilution']
                        and (
                            isinstance(structure['extraction'][0]['sample dilution']['added'], list)
                            or isinstance(structure['extraction'][0]['sample dilution']['added'], dict)
                        )
                    ):
                        # Handle both list and dict for 'added'
                        added = structure['extraction'][0]['sample dilution']['added']
                        if isinstance(added, list) and len(added) > 0:
                            added[0]['value'] = val
                        elif isinstance(added, dict):
                            added['value'] = val
                        print("Extracted sample dilution value for", mapping['new_csv'], ":", val)
                        print("First column header:", mapping.get('first_column_header', ''))
                        found = True
                except Exception:
                    pass
        # Case: part contains "H2O" (e.g., "12mLH2O" or "10mLH2O" or "10H2O")
        elif "H2O" in part:
            # Find the numeric value right before "H2O" in the same string, e.g., "12mLH2O" -> 12
            match = re.search(r'([0-9.]+)\s*[mM][lL]?\s*H2O', part)
            if not match:
                # Also handle cases like "10H2O" (no mL)
                match = re.search(r'([0-9.]+)\s*H2O', part)
            if match:
                try:
                    val = float(match.group(1))
                    if (
                        'extraction' in structure
                        and isinstance(structure['extraction'], list)
                        and len(structure['extraction']) > 0
                        and 'sample dilution' in structure['extraction'][0]
                        and 'added' in structure['extraction'][0]['sample dilution']
                    ):
                        added = structure['extraction'][0]['sample dilution']['added']
                        if isinstance(added, list) and len(added) > 0:
                            added[0]['value'] = val
                        elif isinstance(added, dict):
                            added['value'] = val
                        found = True
                        print(f"Extracted sample dilution value from part '{part}': {val}")
                except Exception:
                    pass
        # Only print if "H2O" is not present in header_parts at all
        if "H2O" not in header_parts and not any("H2O" in p for p in header_parts):
            # If not found, set sample dilution added value to 0.0
            if (
                'extraction' in structure
                and isinstance(structure['extraction'], list)
                and len(structure['extraction']) > 0
                and 'sample dilution' in structure['extraction'][0]
                and 'added' in structure['extraction'][0]['sample dilution']
                and (
                    isinstance(structure['extraction'][0]['sample dilution']['added'], list)
                    or isinstance(structure['extraction'][0]['sample dilution']['added'], dict)
                )
            ):
                added = structure['extraction'][0]['sample dilution']['added']
                if isinstance(added, list) and len(added) > 0:
                    added[0]['value'] = 0.0
                elif isinstance(added, dict):
                    added['value'] = 0.0
                print("No sample dilution value extracted from header_parts:", header_parts)

    # --- Print all values before writing ---
    # Safely extract values from extraction list
    extraction = structure.get("extraction", [])
    if extraction and isinstance(extraction, list):
        extraction0 = extraction[0]
        sample_extracted_value = extraction0.get("sample extracted", {}).get("value", "")
        complexing_agent_added_value = extraction0.get("complexing agent", {}).get("added", {}).get("value", "")
        sample_dilution_added_list = extraction0.get("sample dilution", {}).get("added", [])
        if isinstance(sample_dilution_added_list, list) and sample_dilution_added_list:
            sample_dilution_added_value = sample_dilution_added_list[0].get("value", "")
        else:
            sample_dilution_added_value = ""
    else:
        sample_extracted_value = ""
        complexing_agent_added_value = ""
        sample_dilution_added_value = ""

    # Try to extract sample dilution added value from header_parts if not already set
    if not sample_dilution_added_value:
        for i, part in enumerate(header_parts):
            if part == "H2O" and i > 0:
                val = header_parts[i - 1].replace("mL", "").replace("ml", "")
                try:
                    val = float(val)
                    sample_dilution_added_value = val
                except Exception:
                    sample_dilution_added_value = header_parts[i - 1]
                break
    date_str = structure.get("date", "")
    start_time = None
    if date_str:
        # Find all indices where the date matches in starting_times_dict["date"]
        matching_indices = [i for i, d in enumerate(starting_times_dict["date"]) if d == date_str]
        if matching_indices:
            print(f"Date match found for {date_str} at indices {matching_indices}")
            # Use the first matching index (or handle multiple if needed)
            idx = matching_indices[0]
            start_time = starting_times_dict["starting time"][idx]
            if isinstance(start_time, float) and pd.isna(start_time):
                start_time = None
        if start_time:
            structure["start time"] = start_time

    print(
        f'"{csv_filename}",'
        f'date:"{structure.get("date", "")}",'
        f'irradiation time:"{structure.get("irradiation", {}).get("time", {}).get("value", "")}",'
        f'wavelength"{structure.get("system", {}).get("wavelength", {}).get("value", "")}",'
        f'Loading:"{structure.get("system", {}).get("Loading", {}).get("value", "")}",'
        f'material:"{structure.get("system", {}).get("material", {}).get("name", "")}",'
        f'synthesis date:"{structure.get("system", {}).get("material", {}).get("synthesis date", "")}",'
        f'spot:"{structure.get("spot", "")}",'
        f'batch:"{structure.get("system", {}).get("batch", "")}",'
        f'purging time:"{structure.get("system", {}).get("purging", {}).get("time", {}).get("value", "")}",'
        f'sample extracted:"{sample_extracted_value}",'
        f'TiOSO4"{complexing_agent_added_value}",'
        f'H2O dilution"{sample_dilution_added_value}"'
    )

    # Write the yaml file with preserved indentation and formatting
    yaml_ruamel = YAML()
    yaml_ruamel.indent(mapping=2, sequence=4, offset=2)
    with open(yaml_path, 'w', encoding='utf-8') as out_f:
        yaml_ruamel.dump(structure, out_f)
```
