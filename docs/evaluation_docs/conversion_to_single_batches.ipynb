# Conversion and loading UVV files

```python
import glob
from uvv.package import create_outfiles
```

Select folder from where you would like to convert the files.

```python
input_folder = 'data/raw/'
files = glob.glob(input_folder + '*.csv')
len(files)
# 20240221_UVV_ALLA_LN31o0011_3.csv
# 20240221_UVV_ALLA_LN31o0012_3.csv
```

```python
import os
import traceback

def should_overwrite():
    while True:
        choice = input("Some output files already exist. Overwrite them? (y/n): ").strip().lower()
        if choice in ['y', 'n']:
            return choice == 'y'

outdir = 'data/processed/'
overwrite_decision = None

for csv_file in files:
    output_file = os.path.join(outdir, os.path.basename(csv_file))
    if os.path.exists(output_file):
        if overwrite_decision is None:
            overwrite_decision = should_overwrite()
        if not overwrite_decision:
            print(f"Skipping {csv_file} (output file exists)")
            continue
    print(csv_file)
    try:
        create_outfiles(csv_file=csv_file, outdir=outdir)
    except Exception as e:
        print(f"Error processing {csv_file}: {e}")
```
